```python
def ttl_crea():
    ttl=[]
    ## Définition du référentiel
    ttl.append("""@prefix dct: <http://purl.org/dc/terms/> .
    @prefix eur: <http://data.europa.eu/8mn/euroscivoc/> .
    @prefix fresh: <https://purl.org/fresh/> .
    @prefix term: <https://purl.org/fresh/terminologies/> .
    @prefix mesh: <http://id.nlm.nih.gov/mesh/> .
    @prefix rdf: <http://www.w3.org/1999/02/22-rdf-syntax-ns#> .
    @prefix skos: <http://www.w3.org/2004/02/skos/core#> .
    @prefix vocab: <https://portail-fresh.fr/vocab/> .
    @prefix owl: <http://www.w3.org/2002/07/owl#> .
    @prefix rdfs: <http://www.w3.org/2000/01/rdf-schema#> .
    @prefix pav: <http://purl.org/pav/> .
    @prefix xsd: <http://www.w3.org/2001/XMLSchema#> .

    fresh:terminologies rdf:type owl:Class . """)

    ttl.append(f"""

    fresh:{Nom_terminologie} rdf:type owl:Class ;
        rdfs:subClassOf {Parent} ;
        dct:identifier term:{Code_terminologie} ;
        dct:title '{Titre_FR}'@fr , '{Titre_EN}'@en ;
        dct:description '{Description_FR}'@fr , '{Description_EN}'@en ;
        dct:creator fresh: ;
        pav:createdOn "{CreatedOn}"^^xsd:date ;
        pav:lastUpdatedOn '{LastUpdatedOn}' ;
        

    """)

    md=Metadata.split("|")
    if len(md)>1:
        for el in md:
            ttl.append(f""" 
        fresh:metadataConcerned '{el}';


            """)
    else :
        ttl.append(f""" 
        fresh:metadataConcerned '{Metadata}';


        """)
    ttl.append(f""" 
        dct:license '{License}' .


            """)
    return ttl
```

```python
def ttl_concept(ttl):
    
    for i in range(0, len(df)):
        uri_fresh=df["uri_fresh"][i]
        label_fr=df["label_fr"][i]
        label_en=df["label_en"][i]
        def_fr=df["def_fr"][i]
        def_en=df["def_en"][i]
        exact=df["exact"][i]
        close=df["close"][i]
        broad=df["broad"][i]
        narrow=df["narrow"][i]
        related=df["related"][i]

    
        ttl.append(f"""
        term:{uri_fresh} rdf:type owl:NamedIndividual, fresh:{Nom_terminologie} ;
            skos:prefLabel '{label_fr}'@fr , '{label_en}'@en ;
        """)

        if type(def_fr)==str and type(def_en)==str:
            ttl.append(f"""
            skos:definition '{def_fr}'@fr, '{def_en}'@en ;
        """)
        if type(exact)==str:
            spl=exact.split("|")
            if len(spl)>1:
                 for el in spl:
                    ttl.append(f"""
                                skos:exactMatch '{el}' ;
                            """)    
            else:ttl.append(f"""
            skos:exactMatch '{exact}' ;
            """)
        if type(close)==str:
            spl=close.split("|")
            if len(spl)>1:
                for el in spl:
                    ttl.append(f"""
                                skos:closeMatch '{el}' ;
                            """)    
            else:ttl.append(f"""
            skos:closeMatch '{close}' ;
        """)
        if type(broad)==str:
            spl=broad.split("|")
            if len(spl)>1:
                for el in spl:
                    ttl.append(f"""
                                skos:broadMatch '{el}' ;
                            """)    
            else:ttl.append(f"""
            skos:broadtMatch '{broad}' ;
        """)
        if type(narrow)==str:
            spl=narrow.split("|")
            if len(spl)>1:
                for el in spl:
                    ttl.append(f"""
                                skos:narrowMatch '{el}' ;
                            """)    
            else:ttl.append(f"""
            skos:narrowMatch '{narrow}' ;
        """)
        if type(related)==str:
            spl=related.split("|")
            if len(spl)>1:
                for el in spl:
                    ttl.append(f"""
                                skos:relatedMatch '{el}' ;
                            """)    
            else:ttl.append(f"""
                skos:relatedMatch '{related}' ;
            """)
        ttl.append(f"""
        dct:identifier '{uri_fresh}' .
    """)
    return ttl
```
